```python
import pandas as pd

# Cargar el archivo CSV
df = pd.read_csv("..\\data\\sensores_industriales.csv")
df
```

```python
# 1.- Mostrar la cantidad de registros y de sensores distintos

total_registros = len(df)
total_sensores = df['id_sensor'].nunique()

print("Total de registros:", total_registros)
print("Cantidad de sensores distintos:", total_sensores)
```

```python
# 2.- Calcular la temperatura promedio de cada planta

promedios = df.groupby('planta')['temperatura_c'].mean()

print("Temperatura promedio por planta:")
print(promedios)
```

```python
# 3.- Encontrar la temperatura máxima e identificar el sensor y la fecha correspondientes

temp_max = df['temperatura_c'].max()
print("Temperatura máxima:", temp_max)

# Encontrar el registro con la temperatura máxima
registro_max = df[df['temperatura_c'] == temp_max]
print("Registro con temperatura máxima:")
print(registro_max[['id_sensor', 'fecha_hora']])
```

```python
# 4.- Contar las lecturas con temperatura mayor que 85 °C

lecturas_altas = df[df['temperatura_c'] > 85]
print("Cantidad de lecturas con temperatura mayor que 85 °C:", len(lecturas_altas))
```

```python
# 5.- Identificar la planta con más alertas de temperatura

alertas_por_planta = lecturas_altas['planta'].value_counts()
max_alertas = alertas_por_planta.max()

# Muestra todas las plantas empatadas con el máximo número de alertas
plantas_con_mas_alertas = alertas_por_planta[alertas_por_planta == max_alertas]

print("Planta(s) con más alertas:")
print(plantas_con_mas_alertas)
```

```python
# 6.- Exportar todas las lecturas con alerta a resultados/alertas.csv, conservando las columnas originales

lecturas_alerta = df[df['temperatura_c'] > 85]
lecturas_alerta.to_csv("..//resultados//alertas.csv", index=False)
print("Lecturas con alerta exportadas a resultados//alertas.csv")
```
